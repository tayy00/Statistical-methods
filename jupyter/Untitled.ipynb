```python
#import libraries
#generate random data for every distribution

np.random.seed(40) #seed for making the numbers not change
n = 3000
poissony = np.random.poisson(lam = 100, size = n)
poissony[0:20] #calling the first 20
```

```python
#plot the poisson
plt.hist(poissony, bins = range(0,np.max(poissony)),color = 'red', edgecolor = 'yellow')
plt.title('Number of road accidents in a year - A poisson distribution')
plt.xlabel('Number of road accidents')
plt.ylabel('Frequency')
```

```python
#geometric distrribution
geometry = np.random.geometric(p = 0.4, size = n)
print(geometry)
plt.hist(geometry, bins = range(1, np.max(geometry)+2) , color = 'blue' ,edgecolor = 'black')
plt.title("Number of trials until first success- A geometric distribution")
plt.xlabel('Number of trials')
plt.ylabel('Frequency')
plt.show
```

```python
#binomial distribution
binomy= np.random.binomial(n = 100 , p = 0.15 , size = n)
print(binomy)
#k = 100
plt.hist(binomy, bins = range(0,100),color = 'orange' , edgecolor = 'green')
plt.title('Recoveries of 100 treated - A binomial distribution')
plt.xlabel('Number of recoveries')
plt.ylabel('Frequency')
plt.show()
```
