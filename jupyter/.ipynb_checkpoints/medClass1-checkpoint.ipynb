How each variable relates to health/biomedical data:

Poisson: Counts of rare events (e.g., infections, mutations).

Geometric: Number of trials until first event (e.g., first positive test).

Binomial: Number of successes in fixed trials (e.g., recoveries).

Uniform Discrete: Random assignment (e.g., ward numbers).

Negative Binomial: Trials until a set number of successes (e.g., recoveries).

Uniform Continuous: Random measurement within a range (e.g., test times).

Normal: Biological measurements (e.g., blood pressure).

Exponential: Time between random events (e.g., ER arrivals).

Gamma: Waiting times with shape (e.g., recovery time).

Beta: Proportions (e.g., response rates).

You can run this code in a Python environment with numpy and matplotlib installed to generate and visualize the simulated biomedical data.

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

np.random.seed(42)
n = 1000

# 1. Poisson: Number of new infections per day in a hospital ward (λ=5)
poisson_data = np.random.poisson(lam=5, size=n)

# 2. Geometric: Number of tests until first positive COVID-19 result (p=0.2)
geometric_data = np.random.geometric(p=0.2, size=n)

# 3. Binomial: Number of patients recovering out of 10 treated (p=0.65)
binomial_data = np.random.binomial(n=10, p=0.65, size=n)

# 4. Uniform Discrete: Ward number assigned to new patients (wards 1-5)
uniform_discrete_data = np.random.randint(1, 6, size=n)

# 5. Negative Binomial: Number of patients treated until 5 recoveries (p=0.3)
neg_binomial_data = np.random.negative_binomial(n=5, p=0.3, size=n)

# 6. Uniform Continuous: Time (in minutes) for a blood test result (5-15 min)
uniform_continuous_data = np.random.uniform(5, 15, size=n)

# 7. Normal: Systolic blood pressure in adults (mean=120, std=15)
normal_data = np.random.normal(loc=120, scale=15, size=n)

# 8. Exponential: Time (in minutes) until next ER arrival (mean=8 min)
exponential_data = np.random.exponential(scale=8, size=n)

# 9. Gamma: Time (in days) to recovery from infection (shape=3, scale=4)
gamma_data = np.random.gamma(shape=3, scale=4, size=n)

# 10. Beta: Proportion of patients responding to a new drug (α=2, β=5)
beta_data = np.random.beta(a=2, b=5, size=n)

# Plotting
fig, axs = plt.subplots(5, 2, figsize=(16, 24))
fig.tight_layout(pad=5.0)

# Poisson
axs[0, 0].hist(poisson_data, bins=range(0, np.max(poisson_data)+2), color='blue', edgecolor='black')
axs[0, 0].set_title('Poisson: New Infections per Day')
axs[0, 0].set_xlabel('Number of Patients')
axs[0, 0].set_ylabel('Frequency')

# Geometric
axs[0, 1].hist(geometric_data, bins=range(1, np.max(geometric_data)+2), color='green', edgecolor='black')
axs[0, 1].set_title('Geometric: Tests Until First Positive')
axs[0, 1].set_xlabel('Number of Tests')
axs[0, 1].set_ylabel('Frequency')

# Binomial
axs[1, 0].hist(binomial_data, bins=range(0, 12), color='orange', edgecolor='black')
axs[1, 0].set_title('Binomial: Recoveries out of 10 Treated')
axs[1, 0].set_xlabel('Number of Recoveries')
axs[1, 0].set_ylabel('Frequency')

# Uniform Discrete
axs[1, 1].hist(uniform_discrete_data, bins=np.arange(1, 7)-0.5, color='purple', edgecolor='black')
axs[1, 1].set_title('Uniform Discrete: Ward Number')
axs[1, 1].set_xlabel('Ward Number')
axs[1, 1].set_ylabel('Frequency')

# Negative Binomial
axs[2, 0].hist(neg_binomial_data, bins=range(0, np.max(neg_binomial_data)+2), color='red', edgecolor='black')
axs[2, 0].set_title('Negative Binomial: Patients Treated Until 5 Recoveries')
axs[2, 0].set_xlabel('Number of Patients Treated')
axs[2, 0].set_ylabel('Frequency')

# Uniform Continuous
axs[2, 1].hist(uniform_continuous_data, bins=20, color='cyan', edgecolor='black')
axs[2, 1].set_title('Uniform Continuous: Blood Test Time')
axs[2, 1].set_xlabel('Time (Minutes)')
axs[2, 1].set_ylabel('Frequency')

# Normal
axs[3, 0].hist(normal_data, bins=30, color='magenta', edgecolor='black')
axs[3, 0].set_title('Normal: Systolic Blood Pressure')
axs[3, 0].set_xlabel('Blood Pressure (mmHg)')
axs[3, 0].set_ylabel('Frequency')

# Exponential
axs[3, 1].hist(exponential_data, bins=30, color='yellow', edgecolor='black')
axs[3, 1].set_title('Exponential: Time Until Next ER Arrival')
axs[3, 1].set_xlabel('Time (Minutes)')
axs[3, 1].set_ylabel('Frequency')

# Gamma
axs[4, 0].hist(gamma_data, bins=30, color='brown', edgecolor='black')
axs[4, 0].set_title('Gamma: Time to Recovery')
axs[4, 0].set_xlabel('Time (Days)')
axs[4, 0].set_ylabel('Frequency')

# Beta
axs[4, 1].hist(beta_data, bins=30, color='pink', edgecolor='black')
axs[4, 1].set_title('Beta: Proportion Responding to Drug')
axs[4, 1].set_xlabel('Proportion')
axs[4, 1].set_ylabel('Frequency')

plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import poisson, geom, binom, randint, nbinom, uniform, norm, expon, gamma, beta

# For consistent results
np.random.seed(42)

# Number of samples
n = 1000

# Set up plotting
plt.style.use("seaborn-v0_8-whitegrid")
fig, axs = plt.subplots(5, 2, figsize=(16, 20))
axs = axs.flatten()

# Distribution 1: Poisson - Number of hospital admissions per hour
poisson_data = poisson.rvs(mu=3, size=n)
axs[0].hist(poisson_data, bins=range(0, max(poisson_data)+1), color='skyblue', edgecolor='black')
axs[0].set_title("Poisson: Hospital Admissions per Hour")

# Distribution 2: Geometric - Number of patients until first disease detection
geo_data = geom.rvs(p=0.2, size=n)
axs[1].hist(geo_data, bins=range(1, max(geo_data)+1), color='salmon', edgecolor='black')
axs[1].set_title("Geometric: Screens Until First Positive Case")

# Distribution 3: Binomial - Successful recoveries in a treatment trial
binom_data = binom.rvs(n=10, p=0.6, size=n)
axs[2].hist(binom_data, bins=range(0, 11), color='lightgreen', edgecolor='black')
axs[2].set_title("Binomial: Recoveries in Treatment Trial (n=10, p=0.6)")

# Distribution 4: Discrete Uniform - Patient ID from a batch
uniform_discrete_data = randint.rvs(1000, 1100, size=n)
axs[3].hist(uniform_discrete_data, bins=30, color='plum', edgecolor='black')
axs[3].set_title("Discrete Uniform: Patient ID (1000–1100)")

# Distribution 5: Negative Binomial - Failures before 5 diagnosis successes
nbinom_data = nbinom.rvs(n=5, p=0.4, size=n)
axs[4].hist(nbinom_data, bins=30, color='khaki', edgecolor='black')
axs[4].set_title("Negative Binomial: Failures Before 5 Positive Diagnoses")

# Distribution 6: Uniform (Continuous) - Body temperature in Celsius
uniform_continuous_data = uniform.rvs(loc=36.1, scale=1.2, size=n)
axs[5].hist(uniform_continuous_data, bins=30, color='skyblue', edgecolor='black')
axs[5].set_title("Uniform: Body Temperature (36.1–37.3°C)")

# Distribution 7: Normal - Blood pressure
normal_data = norm.rvs(loc=120, scale=10, size=n)
axs[6].hist(normal_data, bins=30, color='orchid', edgecolor='black')
axs[6].set_title("Normal: Systolic Blood Pressure (mean=120)")

# Distribution 8: Exponential - Time to next ER visit
expon_data = expon.rvs(scale=2, size=n)
axs[7].hist(expon_data, bins=30, color='orange', edgecolor='black')
axs[7].set_title("Exponential: Time Until Next ER Visit")

# Distribution 9: Gamma - Time to recovery
gamma_data = gamma.rvs(a=2, scale=2, size=n)
axs[8].hist(gamma_data, bins=30, color='teal', edgecolor='black')
axs[8].set_title("Gamma: Time to Recovery After Infection")

# Distribution 10: Beta - Proportion of vaccinated individuals
beta_data = beta.rvs(a=2, b=5, size=n)
axs[9].hist(beta_data, bins=30, color='tomato', edgecolor='black')
axs[9].set_title("Beta: Proportion Vaccinated in Population")

# Final layout
for ax in axs:
    ax.set_ylabel("Frequency")
    ax.set_xlabel("Value")

plt.tight_layout()
plt.show()
```

Key Biomedical Applications:
Poisson: Modeling rare events like ER admissions or mutation counts

Geometric: Time-to-event analysis (e.g., trials until success)

Binomial: Clinical trial outcomes with fixed sample sizes

Discrete Uniform: Categorical medical grading systems

Negative Binomial: Overdispersed microbial counts

Continuous Uniform: Equipment measurement ranges

Normal: Biological measurements (BP, cholesterol)

Exponential: Survival analysis with constant hazard

Gamma: Right-skewed duration data (hospital stays)

Beta: Modeling probabilities/proportions (treatment response)

Visualization Features:
Properly labeled axes with medical units

Discrete vs. continuous representation

Kernel density estimates for continuous distributions

Realistic parameter values from medical literature

Clean grid layout for comparison



```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Set style
sns.set(style="whitegrid")
plt.figure(figsize=(15, 18))

# 1. Poisson Distribution (Hospital Admissions)
lambda_ = 5  # Average ER admissions per hour
poisson_data = np.random.poisson(lambda_, 1000)
plt.subplot(5, 2, 1)
sns.histplot(poisson_data, kde=False, discrete=True)
plt.title(f'Poisson (ER Admissions λ={lambda_}/hr)')
plt.xlabel('Patients per hour')

# 2. Geometric Distribution (Clinical Trial Enrollment)
p = 0.1  # Probability of finding eligible patient
geom_data = np.random.geometric(p, 1000)
plt.subplot(5, 2, 2)
sns.histplot(geom_data, kde=False, discrete=True, bins=range(1, 30))
plt.title(f'Geometric (Patients Screened Until Eligibility p={p})')
plt.xlabel('Patients screened')

# 3. Binomial Distribution (Vaccine Effectiveness)
n, p = 100, 0.85  # Patients, efficacy probability
binom_data = np.random.binomial(n, p, 1000)
plt.subplot(5, 2, 3)
sns.histplot(binom_data, kde=False, discrete=True)
plt.title(f'Binomial (Vaccine Efficacy n={n}, p={p})')
plt.xlabel('Protected patients')

# 4. Discrete Uniform (Lab Test Results)
low, high = 1, 6  # Possible severity grades
dunif_data = np.random.randint(low, high+1, 1000)
plt.subplot(5, 2, 4)
sns.histplot(dunif_data, kde=False, discrete=True)
plt.title(f'Discrete Uniform (Disease Severity {low}-{high})')
plt.xlabel('Severity grade')

# 5. Negative Binomial (Pathogen Counts)
r, p = 10, 0.4  # Dispersion, success prob
nbinom_data = np.random.negative_binomial(r, p, 1000)
plt.subplot(5, 2, 5)
sns.histplot(nbinom_data, kde=False, discrete=True)
plt.title(f'Negative Binomial (Bacterial Colonies r={r}, p={p})')
plt.xlabel('Colonies per sample')

# 6. Continuous Uniform (Lab Equipment Calibration)
low, high = 98, 102  # Temperature range (°F)
unif_data = np.random.uniform(low, high, 1000)
plt.subplot(5, 2, 6)
sns.histplot(unif_data, kde=True)
plt.title(f'Uniform (Lab Temp {low}-{high}°F)')
plt.xlabel('Temperature')

# 7. Normal Distribution (Blood Pressure)
mu, sigma = 120, 15  # Mean, std (mmHg)
norm_data = np.random.normal(mu, sigma, 1000)
plt.subplot(5, 2, 7)
sns.histplot(norm_data, kde=True)
plt.title(f'Normal (Systolic BP μ={mu}, σ={sigma})')
plt.xlabel('mmHg')

# 8. Exponential Distribution (Disease-Free Survival)
scale = 5  # Mean survival years
exp_data = np.random.exponential(scale, 1000)
plt.subplot(5, 2, 8)
sns.histplot(exp_data, kde=True)
plt.title(f'Exponential (Disease-Free Survival λ={1/scale})')
plt.xlabel('Years')

# 9. Gamma Distribution (Hospital Stay Duration)
shape, scale = 2, 3  # Parameters
gamma_data = np.random.gamma(shape, scale, 1000)
plt.subplot(5, 2, 9)
sns.histplot(gamma_data, kde=True)
plt.title(f'Gamma (Hospital Stay α={shape}, β={scale})')
plt.xlabel('Days')

# 10. Beta Distribution (Treatment Response Rate)
alpha, beta = 2, 5  # Shape parameters
beta_data = np.random.beta(alpha, beta, 1000)
plt.subplot(5, 2, 10)
sns.histplot(beta_data, kde=True)
plt.title(f'Beta (Drug Response α={alpha}, β={beta})')
plt.xlabel('Response probability')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
from scipy.optimize import minimize
from scipy.stats import norm

# Generate data
np.random.seed(0)
true_mu, true_sigma = 5, 2
data = np.random.normal(true_mu, true_sigma, 1000)

# Negative log-likelihood
def neg_log_likelihood(params):
    mu, sigma = params
    return -np.sum(norm.logpdf(data, mu, sigma))

# Optimization with bounds for σ > 0
initial_guess = [0, 1]
result = minimize(neg_log_likelihood, initial_guess, method='L-BFGS-B', 
                  bounds=[(None, None), (1e-6, None)])

estimated_mu, estimated_sigma = result.x
print(f"μ: {estimated_mu:.3f}, σ: {estimated_sigma:.3f}")
```

```python
from scipy.stats import expon

# Generate data
true_lambda = 0.5
data = np.random.exponential(scale=1/true_lambda, size=1000)

# Negative log-likelihood
def neg_log_likelihood(params):
    lambda_ = params[0]
    return -(len(data) * np.log(lambda_) - lambda_ * np.sum(data))

result = minimize(neg_log_likelihood, [1.0], method='L-BFGS-B', bounds=[(1e-6, None)])
estimated_lambda = result.x[0]
print(f"λ: {estimated_lambda:.3f}")
```

```python
from scipy.stats import poisson

# Generate data
true_lambda = 3
data = np.random.poisson(true_lambda, 1000)

# Negative log-likelihood
def neg_log_likelihood(params):
    lambda_ = params[0]
    return -np.sum(poisson.logpmf(data, lambda_))

result = minimize(neg_log_likelihood, [1.0], method='L-BFGS-B', bounds=[(1e-6, None)])
estimated_lambda = result.x[0]
print(f"λ: {estimated_lambda:.3f}")
```

```python
from scipy.stats import binom

n = 10  # Known number of trials
true_p = 0.3
data = binom.rvs(n, true_p, size=1000)

# Negative log-likelihood
def neg_log_likelihood(params):
    p = params[0]
    return -np.sum(binom.logpmf(data, n, p))

result = minimize(neg_log_likelihood, [0.5], method='L-BFGS-B', bounds=[(0, 1)])
estimated_p = result.x[0]
print(f"p: {estimated_p:.3f}")
```

```python
import numpy as np
from scipy import stats, optimize
import matplotlib.pyplot as plt

def mle_normal(data):
    """Maximum Likelihood Estimation for Normal Distribution"""
    def neg_log_likelihood(params):
        mu, sigma = params
        return -np.sum(stats.norm.logpdf(data, loc=mu, scale=sigma))
    
    # Initial guesses (mean and std deviation)
    init_params = [np.mean(data), np.std(data)]
    
    # Minimize negative log likelihood
    result = optimize.minimize(neg_log_likelihood, init_params, method='L-BFGS-B',
                              bounds=((None, None), (1e-6, None)))
    
    mu_mle, sigma_mle = result.x
    return {'mu': mu_mle, 'sigma': sigma_mle}

def mle_exponential(data):
    """Maximum Likelihood Estimation for Exponential Distribution"""
    def neg_log_likelihood(lambda_param):
        return -np.sum(stats.expon.logpdf(data, scale=1/lambda_param))
    
    # Initial guess (1/mean)
    init_param = [1/np.mean(data)]
    
    # Minimize negative log likelihood
    result = optimize.minimize(neg_log_likelihood, init_param, method='L-BFGS-B',
                              bounds=((1e-6, None),))
    
    lambda_mle = result.x[0]
    return {'lambda': lambda_mle}

def mle_poisson(data):
    """Maximum Likelihood Estimation for Poisson Distribution"""
    def neg_log_likelihood(lambda_param):
        return -np.sum(stats.poisson.logpmf(data, mu=lambda_param))
    
    # Initial guess (mean)
    init_param = [np.mean(data)]
    
    # Minimize negative log likelihood
    result = optimize.minimize(neg_log_likelihood, init_param, method='L-BFGS-B',
                              bounds=((1e-6, None),))
    
    lambda_mle = result.x[0]
    return {'lambda': lambda_mle}

def mle_bernoulli(data):
    """Maximum Likelihood Estimation for Bernoulli Distribution"""
    def neg_log_likelihood(p):
        return -np.sum(stats.bernoulli.logpmf(data, p))
    
    # Initial guess (mean)
    init_param = [np.mean(data)]
    
    # Minimize negative log likelihood
    result = optimize.minimize(neg_log_likelihood, init_param, method='L-BFGS-B',
                              bounds=((1e-6, 1-1e-6),))
    
    p_mle = result.x[0]
    return {'p': p_mle}

def mle_gamma(data):
    """Maximum Likelihood Estimation for Gamma Distribution"""
    def neg_log_likelihood(params):
        alpha, beta = params
        return -np.sum(stats.gamma.logpdf(data, a=alpha, scale=1/beta))
    
    # Method of moments for initial guess
    mean = np.mean(data)
    var = np.var(data)
    init_alpha = mean**2 / var
    init_beta = mean / var
    init_params = [init_alpha, init_beta]
    
    # Minimize negative log likelihood
    result = optimize.minimize(neg_log_likelihood, init_params, method='L-BFGS-B',
                              bounds=((1e-6, None), (1e-6, None)))
    
    alpha_mle, beta_mle = result.x
    return {'alpha': alpha_mle, 'beta': beta_mle}

def mle_geometric(data):
    """Maximum Likelihood Estimation for Geometric Distribution"""
    def neg_log_likelihood(p):
        return -np.sum(stats.geom.logpmf(data, p))
    
    # Initial guess (1/mean)
    init_param = [1/np.mean(data)]
    
    # Minimize negative log likelihood
    result = optimize.minimize(neg_log_likelihood, init_param, method='L-BFGS-B',
                              bounds=((1e-6, 1-1e-6),))
    
    p_mle = result.x[0]
    return {'p': p_mle}

# Generate sample data for testing
np.random.seed(42)

# Normal distribution sample
normal_data = np.random.normal(loc=5, scale=2, size=1000)

# Exponential distribution sample
exp_data = np.random.exponential(scale=1/0.5, size=1000)

# Poisson distribution sample
poisson_data = np.random.poisson(lam=3, size=1000)

# Bernoulli distribution sample
bernoulli_data = np.random.binomial(n=1, p=0.3, size=1000)

# Gamma distribution sample
gamma_data = np.random.gamma(shape=2, scale=1/0.5, size=1000)

# Geometric distribution sample
geom_data = np.random.geometric(p=0.2, size=1000)

# Perform MLE for each distribution
print("Normal Distribution MLE:", mle_normal(normal_data))
print("Exponential Distribution MLE:", mle_exponential(exp_data))
print("Poisson Distribution MLE:", mle_poisson(poisson_data))
print("Bernoulli Distribution MLE:", mle_bernoulli(bernoulli_data))
print("Gamma Distribution MLE:", mle_gamma(gamma_data))
print("Geometric Distribution MLE:", mle_geometric(geom_data))
```

```python
import numpy as np
from scipy.optimize import minimize
from scipy.stats import binom, poisson, norm, expon
import matplotlib.pyplot as plt


np.random.seed(42)
binomial_data = np.random.binomial(n=10, p=0.6, size=100)
poisson_data = np.random.poisson(lam=3, size=100)
normal_data = np.random.normal(loc=5, scale=2, size=100)
exponential_data = np.random.exponential(scale=2, size=100)

def negative_log_likelihood_binomial(p, data, n):
    if not 0 < p[0] < 1:
        return np.inf
    return -np.sum(binom.logpmf(data, n, p[0]))

def negative_log_likelihood_poisson(lmbda, data):
    if lmbda[0] <= 0:
        return np.inf
    return -np.sum(poisson.logpmf(data, lmbda[0]))

def negative_log_likelihood_normal(params, data):
    mu, sigma = params
    if sigma <= 0:
        return np.inf
    return -np.sum(norm.logpdf(data, mu, sigma))

def negative_log_likelihood_exponential(lmbda, data):
    if lmbda[0] <= 0:
        return np.inf
    return -np.sum(expon.logpdf(data, scale=1/lmbda[0]))

# Binomial
n_bin = 10
res_bin = minimize(negative_log_likelihood_binomial, x0=[0.5], args=(binomial_data, n_bin))
print("Binomial MLE for p:", res_bin.x[0])

# Poisson
res_pois = minimize(negative_log_likelihood_poisson, x0=[1], args=(poisson_data,))
print("Poisson MLE for lambda:", res_pois.x[0])

# Normal
res_norm = minimize(negative_log_likelihood_normal, x0=[1, 1], args=(normal_data,))
print("Normal MLE for mu and sigma:", res_norm.x)

# Exponential
res_exp = minimize(negative_log_likelihood_exponential, x0=[1], args=(exponential_data,))
print("Exponential MLE for lambda:", res_exp.x[0])

```

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.stats.diagnostic import het_breuschpagan
from statsmodels.stats.stattools import durbin_watson
from scipy.stats import shapiro

# Sample dataset
# You can replace this with your own dataset
df = sm.datasets.get_rdataset("mtcars", "datasets").data

# Define the independent variables (X) and dependent variable (y)
X = df[['hp', 'wt', 'qsec']]  # independent variables
y = df['mpg']                 # dependent variable

# Add constant to the model (intercept)
X = sm.add_constant(X)

# Fit the model
model = sm.OLS(y, X).fit()

# Print model summary
print(model.summary())

# ---- DIAGNOSTIC TESTS ---- #

# 1. Multicollinearity - VIF
vif_data = pd.DataFrame()
vif_data["feature"] = X.columns
vif_data["VIF"] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]
print("\nVariance Inflation Factors (VIF):")
print(vif_data)

# 2. Residual plots
residuals = model.resid
fitted = model.fittedvalues

plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
sns.residplot(x=fitted, y=residuals, lowess=True, line_kws={'color': 'red'})
plt.xlabel('Fitted values')
plt.ylabel('Residuals')
plt.title('Residuals vs Fitted')

plt.subplot(1, 2, 2)
sns.histplot(residuals, kde=True)
plt.title('Histogram of Residuals')

plt.tight_layout()
plt.show()

# 3. Normality test of residuals
# QQ plot
sm.qqplot(residuals, line='45')
plt.title('QQ Plot')
plt.show()

# Shapiro-Wilk test
shapiro_test = shapiro(residuals)
print("\nShapiro-Wilk Test:")
print(f"Statistic={shapiro_test.statistic:.4f}, p-value={shapiro_test.pvalue:.4f}")

# 4. Homoscedasticity test - Breusch-Pagan
bp_test = het_breuschpagan(residuals, model.model.exog)
print("\nBreusch-Pagan test:")
print(f"LM stat={bp_test[0]:.4f}, p-value={bp_test[1]:.4f}")

# 5. Autocorrelation - Durbin-Watson
dw_stat = durbin_watson(residuals)
print(f"\nDurbin-Watson Statistic: {dw_stat:.4f}")
```

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.stats.diagnostic import het_breuschpagan, linear_rainbow, het_white
from statsmodels.stats.stattools import durbin_watson
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Set random seed for reproducibility
np.random.seed(42)

# Generate synthetic data
n_samples = 200
X1 = np.random.normal(10, 2, n_samples)
X2 = np.random.normal(5, 1, n_samples)
X3 = np.random.normal(8, 1.5, n_samples)
noise = np.random.normal(0, 2, n_samples)

# True coefficients
beta0 = 5
beta1 = 2.5
beta2 = -1.8
beta3 = 0.7

# Generate response variable
y = beta0 + beta1*X1 + beta2*X2 + beta3*X3 + noise

# Create DataFrame
data = pd.DataFrame({'y': y, 'X1': X1, 'X2': X2, 'X3': X3})

# Add a constant for the intercept term
X = sm.add_constant(data[['X1', 'X2', 'X3']])

# Fit the OLS model
model = sm.OLS(data['y'], X)
results = model.fit()

# Print model summary
print("="*80)
print("MODEL SUMMARY")
print("="*80)
print(results.summary())

# Diagnostic tests
print("\n" + "="*80)
print("DIAGNOSTIC TESTS")
print("="*80)

# 1. Normality of residuals (Jarque-Bera test is included in summary)
residuals = results.resid
print("\n1. Normality Tests:")
print(f"Jarque-Bera test (from summary): p-value = {results.jarque_bera[1]:.4f}")

# Additional normality test (Shapiro-Wilk)
shapiro_test = stats.shapiro(residuals)
print(f"Shapiro-Wilk test: p-value = {shapiro_test[1]:.4f}")

# QQ plot for normality check
plt.figure(figsize=(10, 6))
sm.qqplot(residuals, line='s')
plt.title('Q-Q Plot of Residuals')
plt.show()

# 2. Heteroscedasticity tests
print("\n2. Heteroscedasticity Tests:")
# Breusch-Pagan test
bp_test = het_breuschpagan(residuals, results.model.exog)
print(f"Breusch-Pagan test: p-value = {bp_test[1]:.4f}")

# White test
white_test = het_white(residuals, results.model.exog)
print(f"White test: p-value = {white_test[1]:.4f}")

# Plot residuals vs fitted values
plt.figure(figsize=(10, 6))
sns.residplot(x=results.fittedvalues, y=residuals, lowess=True, 
              scatter_kws={'alpha': 0.5}, line_kws={'color': 'red'})
plt.xlabel('Fitted values')
plt.ylabel('Residuals')
plt.title('Residuals vs Fitted Values')
plt.axhline(y=0, color='black', linestyle='--')
plt.show()

# 3. Multicollinearity (VIF)
print("\n3. Multicollinearity (VIF):")
vif_data = pd.DataFrame()
vif_data["Variable"] = X.columns
vif_data["VIF"] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]
print(vif_data)

# 4. Linearity (Rainbow test)
print("\n4. Linearity Test (Rainbow):")
rainbow_test = linear_rainbow(results)
print(f"Rainbow test: p-value = {rainbow_test[1]:.4f}")

# 5. Autocorrelation (Durbin-Watson)
print("\n5. Autocorrelation Test (Durbin-Watson):")
dw_test = durbin_watson(residuals)
print(f"Durbin-Watson statistic: {dw_test:.4f}")
print("Values close to 2 indicate no autocorrelation (0-2: positive, 2-4: negative)")

# 6. Outliers and influence
print("\n6. Outliers and Influence:")
# Cook's distance
influence = results.get_influence()
cooks_distance = influence.cooks_distance[0]

# Plot Cook's distance
plt.figure(figsize=(10, 6))
plt.stem(np.arange(len(cooks_distance)), cooks_distance, markerfmt=",")
plt.title("Cook's Distance")
plt.xlabel("Observation index")
plt.ylabel("Cook's distance")
plt.axhline(y=4/len(residuals), color='red', linestyle='--')  # Common threshold
plt.show()

# High influence points
print(f"\nNumber of observations with Cook's distance > 4/n: {sum(cooks_distance > 4/len(residuals))}")

# 7. Partial regression plots
print("\n7. Partial Regression Plots:")
fig = plt.figure(figsize=(12, 8))
sm.graphics.plot_partregress_grid(results, fig=fig)
plt.tight_layout()
plt.show()

# 8. Leverage plot
print("\n8. Leverage Plot:")
fig, ax = plt.subplots(figsize=(10, 6))
sm.graphics.influence_plot(results, ax=ax, criterion="cooks")
plt.title("Influence Plot")
plt.show()
```
