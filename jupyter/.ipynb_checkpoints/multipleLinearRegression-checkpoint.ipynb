```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import plotly.express as px
from scipy import stats
import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.stats.diagnostic import het_breuschpagan
from statsmodels.stats.stattools import durbin_watson
from scipy.stats import shapiro
import statsmodels.formula.api as smf

%matplotlib inline
```

```python
path = 'global_cancer_patients_2015_2024.csv'
data = pd.read_csv(path)
# data.set_index(data.columns[0], inplace = True)  : code to set a specific column as row names
data.head()
```

```python
# plot distribution of the target variable
plt.figure(figsize=(10, 5))
sns.histplot(data['Target_Severity_Score'], kde = True, bins = 30, color = 'blue')
plt.title('Distribution of Target Severity Score')
plt.xlabel('Severity Score')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(12,7))
sns.heatmap(subset2.corr(), annot = True, fmt = '.2f', cmap = 'viridis')
plt.title('Correlation Matrix')
plt.show()
```

```python
# visualize relationship between alcohol use and target
factors = ['Age', 'Genetic_Risk', 'Air_Pollution', 'Alcohol_Use', 'Smoking', 'Obesity_Level']

for factor in factors:
    plt.figure(figsize=(10, 5))
    sns.scatterplot(data = data, x = factor, y = 'Target_Severity_Score')
    plt.title(f'Severity score vs {factor}')
    plt.xlabel(factor)
    plt.ylabel('Target_Severity_Score')
    plt.show()
    print()
```

```python
# method 1 of building the model
var_factors = ['Genetic_Risk', 'Smoking']
X = data[var_factors]
y = data['Target_Severity_Score']

X_with_constant = sm.add_constant(X)

model = sm.OLS(y, X_with_constant).fit()
print(model.params)
print()
print(model.summary())
```

```python
# method two of building the mlr model
linear_model = smf.ols(formula = 'Target_Severity_Score ~ Genetic_Risk + Smoking', data = data)

# fit the model
lm = linear_model.fit()
print(lm.params)
print()
print(lm.summary())
```

```python
# Calculate VIF for multicollinearity check
print("\nVariance Inflation Factors (VIF):")
vif_data = pd.DataFrame()
vif_data["Variable"] = X_with_constant.columns
vif_data["VIF"] = [variance_inflation_factor(X_with_constant.values, i) for i in range(X_with_constant.shape[1])]
print(vif_data)
```

#### **EXPLANATIONS**
- **R-squared** value of 0.466: This means that about 46.6% of the variability in Target_Severity_Score is explained by Genetic_Risk and Smoking. This is a moderate explanatory power 
- **Adjusted R-squared** of 0.466: implies no overfitting
- **F-statistic** = 2.178e+04, p < 0.001  signifies that the linear model is statistically significant as a whole.
- **Condition Number** of 20.5 suggests no severe multicollinearity between predictors. (Safe threshold is < 30.)
