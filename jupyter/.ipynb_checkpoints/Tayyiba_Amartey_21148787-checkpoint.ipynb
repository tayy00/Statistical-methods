```python
life1['Life.expectancy'] = life1['Life.expectancy'].fillna(life1['Life.expectancy'].mean())
life1['Hepatitis.B'] = life1['Hepatitis.B'].fillna(life1['Hepatitis.B'].mean())
life1['Polio'] = life1['Polio'].fillna(life1['Polio'].mean())
life1['Diphtheria'] = life1['Diphtheria'].fillna(life1['Diphtheria'].mean())


import statsmodels.api as sm

# Select predictor variables and target variable
X = life1[['Hepatitis.B', 'Polio', 'Diphtheria']] 
y = life1['Life.expectancy'] 

# Add constant to the predictors
X = sm.add_constant(X)

# Fit the OLS model
model = sm.OLS(y, X).fit()

# Print results
print(model.params)
print()
print(model.summary())



```

```python
# Calculate VIF for multicollinearity check

print("\n QUESTION 1 H: Variance Inflation Factors (VIF):")
vif_data = pd.DataFrame()
vif_data["Variable"] = X.columns
vif_data["VIF"] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]
print(vif_data)
```

```python
# 2. Residual plots
residuals = model.resid
fitted = model.fittedvalues

plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
sns.residplot(x=fitted, y=residuals, lowess=True, line_kws={'color': 'red'})
plt.xlabel('Fitted values')
plt.ylabel('Residuals')
plt.title('Residuals vs Fitted')

plt.subplot(1, 2, 2)
sns.histplot(residuals, kde=True)
plt.title('Histogram of Residuals')

plt.tight_layout()
plt.show()
```

```python
# 3. Normality test of residuals
# QQ plot
sm.qqplot(residuals, line='45')
plt.title('QQ Plot')
plt.show()

# Shapiro-Wilk test
shapiro_test = shapiro(residuals)
print("\nShapiro-Wilk Test:")
print(f"Statistic={shapiro_test.statistic:.4f}, p-value={shapiro_test.pvalue:.4f}")

# 4. Homoscedasticity test - Breusch-Pagan
bp_test = het_breuschpagan(residuals, model.model.exog)
print("\nBreusch-Pagan test:")
print(f"LM stat={bp_test[0]:.4f}, p-value={bp_test[1]:.4f}")

# 5. Autocorrelation - Durbin-Watson
dw_stat = durbin_watson(residuals)
print(f"\nDurbin-Watson Statistic: {dw_stat:.4f}")
```

Splitting the dataset
- Training 80%
- Validation 10%
- Testing 10%

```python
df_train_full, df_test = train_test_split(cancer1, test_size = 0.1, random_state = 18)
df_train, df_valid = train_test_split(df_train_full, test_size = 1/9, random_state = 18)

print(f'Training dataset {df_train.shape}')
print(f'Validation dataset {df_valid.shape}')
print(f'Testing dataset {df_test.shape}')
```

```python
# select target column
y_train = df_train['binary_diagnosis'].values
y_valid = df_valid['binary_diagnosis'].values
y_test = df_test['binary_diagnosis'].values
```

```python
# delete target columns from the rest of the dataframe
del df_train['binary_diagnosis']
del df_valid['binary_diagnosis']
del df_test['binary_diagnosis']
```
