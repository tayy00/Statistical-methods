```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
import numpy as np
import seaborn as sns
```

```python
air = pd.read_excel('AirQualityUCI.xlsx')
air.head(10)
```

```python
air1 = air.copy()
air1.head(10)
```

```python
print(air1.columns.tolist())
```

```python
air1 = air1.drop(['Date','Time'], axis = 1)
air1.head(10)
```

```python
air2 = air1.corr()
plt.figure(figsize = (20,10))
sns.heatmap(air2, annot = True , cmap = 'viridis')
plt.show()
```

```python
import statsmodels.api as sm
#define dependent variable
Y = air1['CO(GT)']

#define independent variable
X = air1.drop(['CO(GT)'],axis = 1)

# add a constant term for the intercept
X = sm.add_constant(X)

#fit the OLS model (ordinary least squares)
model = sm.OLS(Y,X, missing = 'drop').fit() #missing drop handles any missing values - NaNs

#print the summary
print(model.summary())
```
