How each variable relates to health/biomedical data:

Poisson: Counts of rare events (e.g., infections, mutations).

Geometric: Number of trials until first event (e.g., first positive test).

Binomial: Number of successes in fixed trials (e.g., recoveries).

Uniform Discrete: Random assignment (e.g., ward numbers).

Negative Binomial: Trials until a set number of successes (e.g., recoveries).

Uniform Continuous: Random measurement within a range (e.g., test times).

Normal: Biological measurements (e.g., blood pressure).

Exponential: Time between random events (e.g., ER arrivals).

Gamma: Waiting times with shape (e.g., recovery time).

Beta: Proportions (e.g., response rates).

You can run this code in a Python environment with numpy and matplotlib installed to generate and visualize the simulated biomedical data.

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

np.random.seed(42)
n = 1000

# 1. Poisson: Number of new infections per day in a hospital ward (λ=5)
poisson_data = np.random.poisson(lam=5, size=n)

# 2. Geometric: Number of tests until first positive COVID-19 result (p=0.2)
geometric_data = np.random.geometric(p=0.2, size=n)

# 3. Binomial: Number of patients recovering out of 10 treated (p=0.65)
binomial_data = np.random.binomial(n=10, p=0.65, size=n)

# 4. Uniform Discrete: Ward number assigned to new patients (wards 1-5)
uniform_discrete_data = np.random.randint(1, 6, size=n)

# 5. Negative Binomial: Number of patients treated until 5 recoveries (p=0.3)
neg_binomial_data = np.random.negative_binomial(n=5, p=0.3, size=n)

# 6. Uniform Continuous: Time (in minutes) for a blood test result (5-15 min)
uniform_continuous_data = np.random.uniform(5, 15, size=n)

# 7. Normal: Systolic blood pressure in adults (mean=120, std=15)
normal_data = np.random.normal(loc=120, scale=15, size=n)

# 8. Exponential: Time (in minutes) until next ER arrival (mean=8 min)
exponential_data = np.random.exponential(scale=8, size=n)

# 9. Gamma: Time (in days) to recovery from infection (shape=3, scale=4)
gamma_data = np.random.gamma(shape=3, scale=4, size=n)

# 10. Beta: Proportion of patients responding to a new drug (α=2, β=5)
beta_data = np.random.beta(a=2, b=5, size=n)

# Plotting
fig, axs = plt.subplots(5, 2, figsize=(16, 24))
fig.tight_layout(pad=5.0)

# Poisson
axs[0, 0].hist(poisson_data, bins=range(0, np.max(poisson_data)+2), color='blue', edgecolor='black')
axs[0, 0].set_title('Poisson: New Infections per Day')
axs[0, 0].set_xlabel('Number of Patients')
axs[0, 0].set_ylabel('Frequency')

# Geometric
axs[0, 1].hist(geometric_data, bins=range(1, np.max(geometric_data)+2), color='green', edgecolor='black')
axs[0, 1].set_title('Geometric: Tests Until First Positive')
axs[0, 1].set_xlabel('Number of Tests')
axs[0, 1].set_ylabel('Frequency')

# Binomial
axs[1, 0].hist(binomial_data, bins=range(0, 12), color='orange', edgecolor='black')
axs[1, 0].set_title('Binomial: Recoveries out of 10 Treated')
axs[1, 0].set_xlabel('Number of Recoveries')
axs[1, 0].set_ylabel('Frequency')

# Uniform Discrete
axs[1, 1].hist(uniform_discrete_data, bins=np.arange(1, 7)-0.5, color='purple', edgecolor='black')
axs[1, 1].set_title('Uniform Discrete: Ward Number')
axs[1, 1].set_xlabel('Ward Number')
axs[1, 1].set_ylabel('Frequency')

# Negative Binomial
axs[2, 0].hist(neg_binomial_data, bins=range(0, np.max(neg_binomial_data)+2), color='red', edgecolor='black')
axs[2, 0].set_title('Negative Binomial: Patients Treated Until 5 Recoveries')
axs[2, 0].set_xlabel('Number of Patients Treated')
axs[2, 0].set_ylabel('Frequency')

# Uniform Continuous
axs[2, 1].hist(uniform_continuous_data, bins=20, color='cyan', edgecolor='black')
axs[2, 1].set_title('Uniform Continuous: Blood Test Time')
axs[2, 1].set_xlabel('Time (Minutes)')
axs[2, 1].set_ylabel('Frequency')

# Normal
axs[3, 0].hist(normal_data, bins=30, color='magenta', edgecolor='black')
axs[3, 0].set_title('Normal: Systolic Blood Pressure')
axs[3, 0].set_xlabel('Blood Pressure (mmHg)')
axs[3, 0].set_ylabel('Frequency')

# Exponential
axs[3, 1].hist(exponential_data, bins=30, color='yellow', edgecolor='black')
axs[3, 1].set_title('Exponential: Time Until Next ER Arrival')
axs[3, 1].set_xlabel('Time (Minutes)')
axs[3, 1].set_ylabel('Frequency')

# Gamma
axs[4, 0].hist(gamma_data, bins=30, color='brown', edgecolor='black')
axs[4, 0].set_title('Gamma: Time to Recovery')
axs[4, 0].set_xlabel('Time (Days)')
axs[4, 0].set_ylabel('Frequency')

# Beta
axs[4, 1].hist(beta_data, bins=30, color='pink', edgecolor='black')
axs[4, 1].set_title('Beta: Proportion Responding to Drug')
axs[4, 1].set_xlabel('Proportion')
axs[4, 1].set_ylabel('Frequency')

plt.show()
```

Distribution | Real-World Health Application
Poisson | Emergency room admissions per hour
Geometric | First disease case among screened patients
Binomial | Number of patients who recover after treatment
Discrete Uniform | Random patient ID assignment
Negative Binomial | Trials before multiple successes in diagnostic tests
Uniform (Cont.) | Body temperatures in healthy range
Normal | Blood pressure, height, weight
Exponential | Time between hospital visits or symptom onset
Gamma | Duration of illness or incubation periods
Beta | Proportion outcomes, like vaccination rates

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import poisson, geom, binom, randint, nbinom, uniform, norm, expon, gamma, beta

# For consistent results
np.random.seed(42)

# Number of samples
n = 1000

# Set up plotting
plt.style.use("seaborn-v0_8-whitegrid")
fig, axs = plt.subplots(5, 2, figsize=(16, 20))
axs = axs.flatten()

# Distribution 1: Poisson - Number of hospital admissions per hour
poisson_data = poisson.rvs(mu=3, size=n)
axs[0].hist(poisson_data, bins=range(0, max(poisson_data)+1), color='skyblue', edgecolor='black')
axs[0].set_title("Poisson: Hospital Admissions per Hour")

# Distribution 2: Geometric - Number of patients until first disease detection
geo_data = geom.rvs(p=0.2, size=n)
axs[1].hist(geo_data, bins=range(1, max(geo_data)+1), color='salmon', edgecolor='black')
axs[1].set_title("Geometric: Screens Until First Positive Case")

# Distribution 3: Binomial - Successful recoveries in a treatment trial
binom_data = binom.rvs(n=10, p=0.6, size=n)
axs[2].hist(binom_data, bins=range(0, 11), color='lightgreen', edgecolor='black')
axs[2].set_title("Binomial: Recoveries in Treatment Trial (n=10, p=0.6)")

# Distribution 4: Discrete Uniform - Patient ID from a batch
uniform_discrete_data = randint.rvs(1000, 1100, size=n)
axs[3].hist(uniform_discrete_data, bins=30, color='plum', edgecolor='black')
axs[3].set_title("Discrete Uniform: Patient ID (1000–1100)")

# Distribution 5: Negative Binomial - Failures before 5 diagnosis successes
nbinom_data = nbinom.rvs(n=5, p=0.4, size=n)
axs[4].hist(nbinom_data, bins=30, color='khaki', edgecolor='black')
axs[4].set_title("Negative Binomial: Failures Before 5 Positive Diagnoses")

# Distribution 6: Uniform (Continuous) - Body temperature in Celsius
uniform_continuous_data = uniform.rvs(loc=36.1, scale=1.2, size=n)
axs[5].hist(uniform_continuous_data, bins=30, color='skyblue', edgecolor='black')
axs[5].set_title("Uniform: Body Temperature (36.1–37.3°C)")

# Distribution 7: Normal - Blood pressure
normal_data = norm.rvs(loc=120, scale=10, size=n)
axs[6].hist(normal_data, bins=30, color='orchid', edgecolor='black')
axs[6].set_title("Normal: Systolic Blood Pressure (mean=120)")

# Distribution 8: Exponential - Time to next ER visit
expon_data = expon.rvs(scale=2, size=n)
axs[7].hist(expon_data, bins=30, color='orange', edgecolor='black')
axs[7].set_title("Exponential: Time Until Next ER Visit")

# Distribution 9: Gamma - Time to recovery
gamma_data = gamma.rvs(a=2, scale=2, size=n)
axs[8].hist(gamma_data, bins=30, color='teal', edgecolor='black')
axs[8].set_title("Gamma: Time to Recovery After Infection")

# Distribution 10: Beta - Proportion of vaccinated individuals
beta_data = beta.rvs(a=2, b=5, size=n)
axs[9].hist(beta_data, bins=30, color='tomato', edgecolor='black')
axs[9].set_title("Beta: Proportion Vaccinated in Population")

# Final layout
for ax in axs:
    ax.set_ylabel("Frequency")
    ax.set_xlabel("Value")

plt.tight_layout()
plt.show()
```

Key Biomedical Applications:
Poisson: Modeling rare events like ER admissions or mutation counts

Geometric: Time-to-event analysis (e.g., trials until success)

Binomial: Clinical trial outcomes with fixed sample sizes

Discrete Uniform: Categorical medical grading systems

Negative Binomial: Overdispersed microbial counts

Continuous Uniform: Equipment measurement ranges

Normal: Biological measurements (BP, cholesterol)

Exponential: Survival analysis with constant hazard

Gamma: Right-skewed duration data (hospital stays)

Beta: Modeling probabilities/proportions (treatment response)

Visualization Features:
Properly labeled axes with medical units

Discrete vs. continuous representation

Kernel density estimates for continuous distributions

Realistic parameter values from medical literature

Clean grid layout for comparison



```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Set style
sns.set(style="whitegrid")
plt.figure(figsize=(15, 18))

# 1. Poisson Distribution (Hospital Admissions)
lambda_ = 5  # Average ER admissions per hour
poisson_data = np.random.poisson(lambda_, 1000)
plt.subplot(5, 2, 1)
sns.histplot(poisson_data, kde=False, discrete=True)
plt.title(f'Poisson (ER Admissions λ={lambda_}/hr)')
plt.xlabel('Patients per hour')

# 2. Geometric Distribution (Clinical Trial Enrollment)
p = 0.1  # Probability of finding eligible patient
geom_data = np.random.geometric(p, 1000)
plt.subplot(5, 2, 2)
sns.histplot(geom_data, kde=False, discrete=True, bins=range(1, 30))
plt.title(f'Geometric (Patients Screened Until Eligibility p={p})')
plt.xlabel('Patients screened')

# 3. Binomial Distribution (Vaccine Effectiveness)
n, p = 100, 0.85  # Patients, efficacy probability
binom_data = np.random.binomial(n, p, 1000)
plt.subplot(5, 2, 3)
sns.histplot(binom_data, kde=False, discrete=True)
plt.title(f'Binomial (Vaccine Efficacy n={n}, p={p})')
plt.xlabel('Protected patients')

# 4. Discrete Uniform (Lab Test Results)
low, high = 1, 6  # Possible severity grades
dunif_data = np.random.randint(low, high+1, 1000)
plt.subplot(5, 2, 4)
sns.histplot(dunif_data, kde=False, discrete=True)
plt.title(f'Discrete Uniform (Disease Severity {low}-{high})')
plt.xlabel('Severity grade')

# 5. Negative Binomial (Pathogen Counts)
r, p = 10, 0.4  # Dispersion, success prob
nbinom_data = np.random.negative_binomial(r, p, 1000)
plt.subplot(5, 2, 5)
sns.histplot(nbinom_data, kde=False, discrete=True)
plt.title(f'Negative Binomial (Bacterial Colonies r={r}, p={p})')
plt.xlabel('Colonies per sample')

# 6. Continuous Uniform (Lab Equipment Calibration)
low, high = 98, 102  # Temperature range (°F)
unif_data = np.random.uniform(low, high, 1000)
plt.subplot(5, 2, 6)
sns.histplot(unif_data, kde=True)
plt.title(f'Uniform (Lab Temp {low}-{high}°F)')
plt.xlabel('Temperature')

# 7. Normal Distribution (Blood Pressure)
mu, sigma = 120, 15  # Mean, std (mmHg)
norm_data = np.random.normal(mu, sigma, 1000)
plt.subplot(5, 2, 7)
sns.histplot(norm_data, kde=True)
plt.title(f'Normal (Systolic BP μ={mu}, σ={sigma})')
plt.xlabel('mmHg')

# 8. Exponential Distribution (Disease-Free Survival)
scale = 5  # Mean survival years
exp_data = np.random.exponential(scale, 1000)
plt.subplot(5, 2, 8)
sns.histplot(exp_data, kde=True)
plt.title(f'Exponential (Disease-Free Survival λ={1/scale})')
plt.xlabel('Years')

# 9. Gamma Distribution (Hospital Stay Duration)
shape, scale = 2, 3  # Parameters
gamma_data = np.random.gamma(shape, scale, 1000)
plt.subplot(5, 2, 9)
sns.histplot(gamma_data, kde=True)
plt.title(f'Gamma (Hospital Stay α={shape}, β={scale})')
plt.xlabel('Days')

# 10. Beta Distribution (Treatment Response Rate)
alpha, beta = 2, 5  # Shape parameters
beta_data = np.random.beta(alpha, beta, 1000)
plt.subplot(5, 2, 10)
sns.histplot(beta_data, kde=True)
plt.title(f'Beta (Drug Response α={alpha}, β={beta})')
plt.xlabel('Response probability')

plt.tight_layout()
plt.show()
```
